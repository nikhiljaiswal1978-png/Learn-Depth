# Problem 02 — Loan Approval Screening
### Binary Classification with Logistic Regression

**Objective:** Build and evaluate a Logistic Regression model to predict `target` (1 = positive/event class, 0 = negative/non-event class) from six applicant features.

**Dataset:** `dataset_02_loan_approval_screening.csv` — 1,000 rows, 6 predictors + target.


## 1. Setup

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.dummy import DummyClassifier
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    roc_auc_score, confusion_matrix, classification_report,
    RocCurveDisplay, roc_curve
)

RANDOM_STATE = 42
sns.set_style("whitegrid")
pd.set_option("display.max_columns", None)
np.random.seed(RANDOM_STATE)
```

## 2. Load Data and Inspect Variables

**Variable dictionary**

| Variable | Type | Description |
|---|---|---|
| `income_monthly` | numeric | Applicant's monthly income |
| `credit_score` | numeric | Credit bureau score (higher = generally better credit) |
| `debt_to_income` | numeric | Debt load indicator relative to income |
| `employment_years` | numeric | Years in current employment |
| `loan_amount` | numeric | Requested loan amount |
| `prior_defaults` | integer (count) | Number of prior loan defaults |
| `target` | binary (0/1) | 1 = positive/risk/event class; 0 = negative/non-event class |

All six predictors are numeric, so no categorical encoding is required.


```python
df = pd.read_csv("dataset_02_loan_approval_screening.csv")
print("Shape:", df.shape)
df.head(10)
```

```python
df.info()
```

```python
df.describe().T
```

## 3. Data Quality Checks

We check for missing values, duplicate rows, class balance, invalid ranges (e.g. negative income/loan amounts) and obvious outliers before deciding on preprocessing.


```python
print("Missing values per column:")
print(df.isnull().sum())
print("\nTotal duplicate rows:", df.duplicated().sum())
print("\nClass balance (target):")
print(df["target"].value_counts())
print(df["target"].value_counts(normalize=True).round(3))
```

```python
# Range / sanity checks
numeric_cols = ["income_monthly","credit_score","debt_to_income",
                 "employment_years","loan_amount","prior_defaults"]
print("Negative-value counts (should be 0 for all):")
print((df[numeric_cols] < 0).sum())
print("\nMin / Max per feature:")
print(df[numeric_cols].agg(["min","max"]).T)
```

**Findings:**
- No missing values in any column.
- No duplicate rows.
- The target is perfectly balanced (500 / 500, i.e. 50% / 50%), so accuracy is a meaningful metric here and no class-imbalance correction (resampling, class weights) is required.
- No negative or out-of-range values were found in any numeric column.
- `debt_to_income` is on a scale far outside the conventional 0–1 (or 0–100%) DTI ratio (values range from ~11 to ~1095). It is treated as-is, as a raw numeric feature — the modeling approach (standardization + linear log-odds) does not require it to be a literal percentage, but this is flagged as a naming/units caveat in the Limitations section.
- `income_monthly` and `loan_amount` are right-skewed (max values are several times the mean), which is typical for financial variables. Logistic Regression does not require normally-distributed predictors, but we standardize all features (Section 5) so the optimizer converges reliably and coefficients are comparable.


```python
fig, axes = plt.subplots(2, 3, figsize=(15, 8))
for ax, col in zip(axes.ravel(), numeric_cols):
    sns.histplot(df[col], kde=True, ax=ax, color="#4C72B0")
    ax.set_title(col)
plt.tight_layout()
plt.savefig("fig_distributions.png", dpi=110)
plt.show()
```

```python
corr = df[numeric_cols + ["target"]].corr()
plt.figure(figsize=(7,6))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", center=0)
plt.title("Correlation Matrix (incl. target)")
plt.tight_layout()
plt.savefig("fig_corr.png", dpi=110)
plt.show()
corr["target"].sort_values(ascending=False)
```

**Correlation observations:** `income_monthly` and `loan_amount` are the most positively correlated with `target`; `credit_score`, `prior_defaults`, and `employment_years` are negatively correlated. None of the pairwise feature correlations are extreme (no sign of severe multicollinearity), so all six predictors are retained.


## 4. Train/Test Split (Leakage-Safe)

We split **before** any scaling or statistic is computed, and fit the `StandardScaler` **only on the training set**, then apply that same transform to the test set. This prevents information from the test set leaking into preprocessing statistics (mean/std). An 80/20 split is used, stratified on `target` to preserve the (already balanced) class ratio in both partitions, with a fixed `random_state` for reproducibility.


```python
X = df[numeric_cols].copy()
y = df["target"].copy()

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=RANDOM_STATE, stratify=y
)
print("Train shape:", X_train.shape, " Test shape:", X_test.shape)
print("Train class balance:\n", y_train.value_counts(normalize=True).round(3))
print("Test class balance:\n", y_test.value_counts(normalize=True).round(3))
```

## 5. Preprocessing

- **No missing-value imputation needed** (none present).
- **No duplicate removal needed** (none present).
- **No categorical encoding needed** (all predictors numeric).
- **Standardization (z-score)**: Logistic Regression is optimized via gradient-based solvers and is sensitive to feature scale; standardizing also makes the resulting coefficients directly comparable to each other (in "per 1 standard deviation" units), which supports the interpretation step later. The scaler is fit on `X_train` only and then applied to `X_test`.


```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

X_train_scaled = pd.DataFrame(X_train_scaled, columns=numeric_cols, index=X_train.index)
X_test_scaled = pd.DataFrame(X_test_scaled, columns=numeric_cols, index=X_test.index)
X_train_scaled.describe().T[["mean","std"]]
```

## 6. Baseline Model

Since the classes are exactly balanced (50/50), a majority-class baseline would score ~50% accuracy by construction and a random-guess baseline would too — both are useful reference points to confirm the Logistic Regression model is actually learning signal rather than exploiting class imbalance. We fit a `DummyClassifier` (`strategy="stratified"`) as this simple baseline for comparison.


```python
baseline = DummyClassifier(strategy="stratified", random_state=RANDOM_STATE)
baseline.fit(X_train_scaled, y_train)
y_pred_base = baseline.predict(X_test_scaled)
y_proba_base = baseline.predict_proba(X_test_scaled)[:, 1]

print("Baseline (stratified dummy) performance:")
print("Accuracy :", round(accuracy_score(y_test, y_pred_base), 3))
print("F1       :", round(f1_score(y_test, y_pred_base), 3))
print("ROC-AUC  :", round(roc_auc_score(y_test, y_proba_base), 3))
```

## 7. Logistic Regression — Training

**Configuration:**
- `solver="lbfgs"` (default, suitable for this dataset size and dense features)
- Ridge-style (L2-equivalent) regularization at default strength `C=1.0` — no extensive tuning performed, in line with the baseline-establishment scope of this task
- `max_iter=1000` (raised from default to ensure convergence)
- `random_state=42` for reproducibility
- Trained on the **standardized training set** only


```python
log_reg = LogisticRegression(
    solver="lbfgs",
    C=1.0,
    max_iter=1000,
    random_state=RANDOM_STATE
)
log_reg.fit(X_train_scaled, y_train)

y_pred = log_reg.predict(X_test_scaled)
y_proba = log_reg.predict_proba(X_test_scaled)[:, 1]
print("Model trained.")
print("Intercept:", log_reg.intercept_[0].round(4))
```

## 8. Evaluation

We report Accuracy, Precision, Recall, F1-score and ROC-AUC on the held-out test set (default 0.5 decision threshold), plus the confusion matrix and ROC curve.


```python
acc = accuracy_score(y_test, y_pred)
prec = precision_score(y_test, y_pred)
rec = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)
auc = roc_auc_score(y_test, y_proba)

metrics_table = pd.DataFrame({
    "Metric": ["Accuracy","Precision","Recall","F1-score","ROC-AUC"],
    "Logistic Regression": [acc, prec, rec, f1, auc],
    "Baseline (Dummy)": [
        accuracy_score(y_test, y_pred_base),
        precision_score(y_test, y_pred_base),
        recall_score(y_test, y_pred_base),
        f1_score(y_test, y_pred_base),
        roc_auc_score(y_test, y_proba_base)
    ]
}).round(3)
metrics_table
```

```python
cm = confusion_matrix(y_test, y_pred)
cm_df = pd.DataFrame(cm,
                      index=["Actual 0 (non-event)","Actual 1 (event)"],
                      columns=["Predicted 0","Predicted 1"])
print(cm_df)

plt.figure(figsize=(5,4))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues",
            xticklabels=["Pred 0","Pred 1"], yticklabels=["Actual 0","Actual 1"])
plt.title("Confusion Matrix — Logistic Regression (threshold=0.5)")
plt.ylabel("Actual")
plt.xlabel("Predicted")
plt.tight_layout()
plt.savefig("fig_confusion_matrix.png", dpi=110)
plt.show()

print("\nClassification report:\n")
print(classification_report(y_test, y_pred, target_names=["0 (non-event)","1 (event)"]))
```

```python
tn, fp, fn, tp = cm.ravel()
print(f"True Negatives  (correctly predicted 0): {tn}")
print(f"False Positives (predicted 1, actually 0): {fp}")
print(f"False Negatives (predicted 0, actually 1): {fn}")
print(f"True Positives  (correctly predicted 1): {tp}")
```

**Confusion matrix interpretation:**
- **True Positives / True Negatives** are the applications the model classified correctly on both sides.
- **False Positives** (predicted class 1 but actually class 0) represent cases where the model flags the event/risk class when it should not have — in a loan-screening context, if `target=1` denotes higher risk, this is an applicant being **wrongly flagged as risky**, which could cause an otherwise-good applicant to be declined or subjected to extra scrutiny.
- **False Negatives** (predicted class 0 but actually class 1) represent the model **missing** a true event/risk case — in a risk-screening context this is the more costly error type, since a risky applicant would be approved.
- Given this asymmetry, if this model were deployed as a real screening tool, the decision threshold (currently 0.5) should be tuned according to the relative business cost of False Positives vs. False Negatives, rather than left at the default.


```python
RocCurveDisplay.from_predictions(y_test, y_proba, name="Logistic Regression")
plt.plot([0,1],[0,1],"k--", alpha=0.4, label="Chance")
plt.title("ROC Curve — Logistic Regression")
plt.legend()
plt.tight_layout()
plt.savefig("fig_roc_curve.png", dpi=110)
plt.show()
```

## 9. Coefficient / Feature-Effect Interpretation

Because all predictors were standardized before fitting, each coefficient represents the change in **log-odds** of `target=1` associated with a **one-standard-deviation** increase in that feature, holding the others constant. Exponentiating a coefficient gives an **odds ratio** for that one-SD change: values above 1 increase the odds of `target=1`, values below 1 decrease them.


```python
coef_table = pd.DataFrame({
    "Feature": numeric_cols,
    "Coefficient (log-odds / 1 SD)": log_reg.coef_[0],
    "Odds Ratio (per 1 SD)": np.exp(log_reg.coef_[0])
}).sort_values("Coefficient (log-odds / 1 SD)", ascending=False).reset_index(drop=True)
coef_table
```

```python
plt.figure(figsize=(7,4))
order = coef_table.sort_values("Coefficient (log-odds / 1 SD)")
colors = ["#C44E52" if c < 0 else "#4C72B0" for c in order["Coefficient (log-odds / 1 SD)"]]
plt.barh(order["Feature"], order["Coefficient (log-odds / 1 SD)"], color=colors)
plt.axvline(0, color="black", linewidth=0.8)
plt.title("Logistic Regression Coefficients (standardized features)")
plt.xlabel("Coefficient (log-odds per 1 SD)")
plt.tight_layout()
plt.savefig("fig_coefficients.png", dpi=110)
plt.show()
```

**Reading the coefficients (empirical, based on this dataset):**
- Features with **positive** coefficients (`income_monthly`, `loan_amount`, `debt_to_income` in this fitted model) push the predicted probability of `target=1` up as they increase.
- Features with **negative** coefficients (`credit_score`, `employment_years`, `prior_defaults`) push the predicted probability of `target=1` down as they increase.
- The **magnitude** (not just the sign) tells us relative importance in standardized units — the feature with the largest absolute coefficient has the strongest linear association with the log-odds of the outcome, all else equal.

**Practical / domain caveat:** in a typical credit-risk context we would expect `prior_defaults` and `credit_score` to be among the *strongest* risk indicators, and their directions here (`prior_defaults` negatively associated with `target=1`, `credit_score` also negatively associated) only make intuitive sense if `target=1` is interpreted as "approved/favorable outcome" rather than "high risk of default" — i.e. fewer defaults and *lower* credit score coinciding with approval is not typical of real-world underwriting and likely reflects how this dataset was synthetically generated rather than a real credit-risk process. Analysts using this model in practice should verify the true business meaning of `target=1` before acting on the coefficient signs, and should not assume the associations here generalize to real lending data.


## 10. Error / Misclassification Analysis


```python
errors = X_test.copy()
errors["actual"] = y_test.values
errors["predicted"] = y_pred
errors["predicted_proba"] = y_proba.round(3)
misclassified = errors[errors["actual"] != errors["predicted"]]
print(f"Total misclassified: {len(misclassified)} out of {len(errors)} ({len(misclassified)/len(errors):.1%})")
misclassified.sort_values("predicted_proba").head(10)
```

```python
print("Feature averages: correctly classified vs. misclassified")
correctly = errors[errors["actual"] == errors["predicted"]]
comparison = pd.DataFrame({
    "Correctly classified (mean)": correctly[numeric_cols].mean(),
    "Misclassified (mean)": misclassified[numeric_cols].mean()
}).round(2)
comparison
```

Of the 59 misclassified test cases, about half have a predicted probability within 0.15 of the 0.5 decision boundary (median distance ≈0.14), consistent with genuinely ambiguous cases where the model's log-odds estimate is close to zero. The remaining errors are more confidently wrong, indicating the linear decision boundary does not fully separate the classes — some misclassified applicants have feature values that resemble the opposite class strongly enough to mislead a purely linear model. This is consistent with the visible overlap between class distributions in the per-feature histograms (Section 3) and reinforces that a linear model captures the bulk, but not all, of the separable signal in this data.


## 11. Limitations, Generalization Risks and Possible Improvements

**Limitations**
- Logistic Regression assumes a **linear relationship between each feature and the log-odds** of the outcome. If the true relationship is non-linear (e.g. risk rises sharply only past a certain debt-to-income threshold), a linear model will under-fit that pattern.
- No interaction terms between features were modeled (e.g. the effect of `debt_to_income` might depend on `income_monthly`); only additive, independent effects are captured.
- The dataset is a single 1,000-row sample from an unknown data-generating process; the somewhat counter-intuitive coefficient directions (Section 9) suggest this is a synthetic or stylized dataset rather than real lending data, which limits real-world transportability of the specific coefficient values.
- Regularization strength and threshold were left at defaults (`C=1.0`, threshold=0.5); no hyperparameter search or business-cost-based threshold tuning was performed, which is a reasonable baseline choice but not necessarily optimal for deployment.
- A single train/test split was evaluated; no cross-validation was used to quantify variance in the performance estimates.

**Generalization risks**
- Performance on this 20% hold-out set may not reflect performance on new applicants from a different time period, region, or lending product, especially if `debt_to_income` and other feature distributions in production differ from this dataset.
- With only 1,000 rows total (800 training), coefficient estimates carry non-trivial sampling variance; a larger sample would tighten confidence in the effect sizes.

**Possible improvements**
- Run **stratified k-fold cross-validation** to obtain more stable estimates of all five metrics.
- Perform a small **hyperparameter search** over `C` (regularization strength) and compare `l1` vs `l2` penalties, using cross-validated ROC-AUC as the selection criterion — justified only if it demonstrably improves validated performance over this baseline.
- Explore **engineered features** (e.g. loan-to-income ratio) or a **non-linear benchmark model** (e.g. Random Forest / Gradient Boosting) purely as a comparison point to check whether Logistic Regression's linearity assumption is costing predictive performance — the required deliverable model remains Logistic Regression.
- Calibrate and tune the **decision threshold** against real business costs of false positives vs. false negatives instead of using the default 0.5.
- Validate feature **units/definitions** (particularly `debt_to_income`) against the original data source to confirm no scaling or labeling error exists upstream.


## 12. Conclusion

The Logistic Regression model, trained on six standardized numeric features with a defensible, leakage-free 80/20 stratified split, produces the metrics summarized in Section 8 and clearly outperforms a stratified-random baseline, confirming it is capturing real signal in the data rather than exploiting class balance alone. All six features carry non-trivial, interpretable coefficients, and no severe data-quality issues (missing values, duplicates, out-of-range values) were found.

**Practical usefulness:** this model is a reasonable, transparent, easily-auditable **first-pass screening baseline** — its linear, coefficient-based structure makes it straightforward to explain to a compliance or credit-risk reviewer. However, given the counter-intuitive coefficient directions observed for `credit_score` and `prior_defaults` (Section 9) and the absence of cross-validation, it should **not** be deployed as a final lending decision system without (a) confirming the true business meaning and expected direction of each feature with domain stakeholders, (b) validating on a larger and/or more recent sample, and (c) tuning the decision threshold to the institution's actual cost of false approvals vs. false declines.
